```python
"""
Data title: Occupational Safety and Health Administration (OSHA) Injury Tracking Application (ITA) from 2020 to 2023

Data source: https://www.osha.gov/injuryreporting/ita/

Data selection: The dataset is ideal for machine learning analysis due to the diverse structured and unstructured data elements related to workplace safety. It includes categorical and numerical variables like NAICS codes, establishment size, and incident outcomes, as well as narrative text fields describing incidents and contributing factors, making it suitable for natural language processing. The dataset providing a rich foundation for both supervised and unsupervised learning. Its volume and variability support robust training, validation, and testing of models.

Data description: The dataset contains detailed records of workplace injuries and illnesses reported to OSHA from 2020 to 2023. Each row represents a single incident and includes both structured data (e.g., incident date, injury type, establishment size, NAICS industry code, job title, number of days away from work) and unstructured narrative text describing the circumstances of the event. Key identifiers include establishment and case numbers, as well as Standard Occupational Classification (SOC) codes assigned using machine learning. From OSHA: "The NIOCCS system uses machine learning, a form of Artificial Intelligence (AI), to assign SOC codes, SOC descriptions, and SOC probability for ITA cases"

--33 different data elements: id, company_name, establishment_name, ein, street_address, city, state, zip_code, naics_code, industry_description, annual_average_employees, total_hours_worked, no_injuries_illnesses, total_deaths, total_dafw_cases, total_djtr_cases, total_other_cases, total_dafw_days, total_djtr_days, total_injuries, total_poisonings, total_respiratory_conditions, total_skin_disorders, total_hearing_loss, total_other_illnesses, establishment_id, establishment_type, size, year_filing_for, NAICS ID, Industry Name, NAICS ID2, IndustryName2

"""
```

```python
#Data preparation

#--Uploading data
df = spark.read.csv("/Volumes/workspace/default/final/OSHA Injury Tracking Application 2020-2023.csv", header=True, inferSchema=True)
df.display()
```

```python
from pyspark.sql.functions import length, lit, concat, isnan, col, lower, trim, regexp_extract, expr, regexp_replace

#Evaluate data to convert based on description from OSHA
df.printSchema()
```

```python
#Preparing data to clean
df = df.withColumn("id", regexp_replace("id", r"\.00$", ""))
df = df.withColumn("id", col("id").cast("int"))

df = df.withColumn("naics_code", regexp_replace("naics_code", r"\.00$", ""))
df = df.withColumn("naics_code", col("naics_code").cast("int"))

df = df.withColumn("annual_average_employees", regexp_replace("annual_average_employees", r"\.00$", ""))
df = df.withColumn("annual_average_employees", col("annual_average_employees").cast("int"))

df = df.withColumn("total_hours_worked", regexp_replace("total_hours_worked", r"\.00$", ""))
df = df.withColumn("total_hours_worked", col("total_hours_worked").cast("bigint"))

df = df.withColumn("no_injuries_illnesses", regexp_replace("no_injuries_illnesses", r"\.00$", ""))
df = df.withColumn("no_injuries_illnesses", col("no_injuries_illnesses").cast("int"))

df = df.withColumn("total_deaths", regexp_replace("total_deaths", r"\.00$", ""))
df = df.withColumn("total_deaths", col("total_deaths").cast("int"))

df = df.withColumn("establishment_id", regexp_replace("establishment_id", r"\.00$", ""))
df = df.withColumn("establishment_id", col("establishment_id").cast("int"))

df = df.withColumn("establishment_type", regexp_replace("establishment_type", r"\.00$", ""))
df = df.withColumn("establishment_type", col("establishment_type").cast("int"))

df = df.withColumn("size", regexp_replace("size", r"\.00$", ""))
df = df.withColumn("size", col("size").cast("int"))

df.printSchema()
```

```python
#Clean data

#--Remove rows without IDs
df = df.dropna(subset=["id"])

#--Only use valid IDs
df = df.filter(
    (col("id").isNotNull()) &
    (~isnan(col("id"))) &
    (length(concat(col("id"), lit(""))) <= 7)
)

#--Normalization
df = df.withColumn("id", trim(col("id")))
df = df.withColumn("company_name", lower(trim(col("company_name"))))
df = df.withColumn("establishment_name", lower(trim(col("establishment_name"))))
df = df.withColumn("ein", trim(col("ein")))
df = df.withColumn("city", lower(trim(col("city"))))
df = df.withColumn("state", lower(trim(col("state"))))
df = df.withColumn("industry_description", lower(trim(col("industry_description"))))
df = df.withColumn("Industry Name", lower(trim(col("Industry Name"))))
df = df.withColumn("Industry Name2", lower(trim(col("Industry Name2"))))
df = df.withColumn("zip_code", regexp_extract(col("zip_code"), r"^(\d{5})", 1))

df = df.withColumnRenamed("NAICS ID", "NAICS_ID")
df = df.withColumnRenamed("Industry Name", "Industry_Name")
df = df.withColumnRenamed("NAICS ID2", "NAICS_ID2")
df = df.withColumnRenamed("Industry Name2", "Industry_Name2")

df.display()
```
